```python
import os
import zipfile
import urllib.request

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split, GridSearchCV, StratifiedKFold
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    classification_report,
    ConfusionMatrixDisplay
)

from sklearn.inspection import permutation_importance
import joblib

RANDOM_STATE = 42
```

```python
import os
import zipfile
import urllib.request
import pandas as pd
from io import BytesIO

url = "https://archive.ics.uci.edu/static/public/222/bank+marketing.zip"

os.makedirs("/content/data", exist_ok=True)

zip_path = "/content/data/bank-marketing.zip"

urllib.request.urlretrieve(url, zip_path)

def find_csv(zip_data):
    with zipfile.ZipFile(zip_data, "r") as z:

        for name in z.namelist():
            if os.path.basename(name).lower() == "bank-full.csv":
                return z.read(name)

        for name in z.namelist():
            if name.lower().endswith(".zip"):
                nested_data = z.read(name)
                result = find_csv(BytesIO(nested_data))

                if result is not None:
                    return result

    return None


with open(zip_path, "rb") as f:
    csv_data = find_csv(BytesIO(f.read()))

if csv_data is None:
    raise FileNotFoundError("bank-full.csv not found")

df = pd.read_csv(BytesIO(csv_data), sep=";")

print("Dataset loaded successfully!")
print("Shape:", df.shape)
display(df.head())
```

```python
print(df.info())
print(df.isnull().sum())
print(df["y"].value_counts())
```

```python
df["y"].value_counts().plot(
    kind="bar",
    title="Target Distribution"
)

plt.xlabel("Subscription")
plt.ylabel("Count")
plt.show()
```

```python
df = df.drop(columns=["duration"])

df["y"] = df["y"].map({
    "yes": 1,
    "no": 0
})

X = df.drop(columns=["y"])
y = df["y"]

print(X.shape)
print(y.shape)
```

```python
numeric_columns = X.select_dtypes(
    include=["int64", "float64"]
).columns

X[numeric_columns].hist(
    figsize=(14, 10),
    bins=30
)

plt.tight_layout()
plt.show()
```

```python
categorical_columns = X.select_dtypes(
    include=["object"]
).columns

for column in categorical_columns:
    rates = df.groupby(column)["y"].mean().sort_values()

    rates.plot(
        kind="bar",
        figsize=(9, 4),
        title=f"Subscription Rate by {column}"
    )

    plt.ylabel("Subscription Rate")
    plt.tight_layout()
    plt.show()
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=RANDOM_STATE,
    stratify=y
)

print("Train:", X_train.shape)
print("Test:", X_test.shape)
```

```python
numeric_features = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X.select_dtypes(
    include=["object"]
).columns.tolist()

numeric_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("numeric", numeric_transformer, numeric_features),
    ("categorical", categorical_transformer, categorical_features)
])
```

```python
models = {
    "Logistic Regression": (
        LogisticRegression(
            max_iter=2000,
            class_weight="balanced"
        ),
        {
            "model__C": [0.01, 0.1, 1, 10]
        }
    ),

    "Decision Tree": (
        DecisionTreeClassifier(
            random_state=RANDOM_STATE,
            class_weight="balanced"
        ),
        {
            "model__max_depth": [5, 10, 20, None],
            "model__min_samples_leaf": [1, 5, 10]
        }
    ),

    "Random Forest": (
        RandomForestClassifier(
            random_state=RANDOM_STATE,
            class_weight="balanced_subsample",
            n_jobs=-1
        ),
        {
            "model__n_estimators": [100, 200],
            "model__max_depth": [10, 20, None],
            "model__min_samples_leaf": [1, 5]
        }
    )
}

cv = StratifiedKFold(
    n_splits=3,
    shuffle=True,
    random_state=RANDOM_STATE
)

results = []
best_score = 0
best_model = None
best_name = None

for name, (model, parameters) in models.items():

    pipeline = Pipeline([
        ("preprocessor", preprocessor),
        ("model", model)
    ])

    grid = GridSearchCV(
        pipeline,
        parameters,
        scoring="roc_auc",
        cv=cv,
        n_jobs=1,
        verbose=1
    )

    grid.fit(X_train, y_train)

    results.append({
        "Model": name,
        "CV ROC-AUC": grid.best_score_,
        "Best Parameters": grid.best_params_
    })

    if grid.best_score_ > best_score:
        best_score = grid.best_score_
        best_model = grid.best_estimator_
        best_name = name

results_df = pd.DataFrame(results)
results_df
```

```python
y_pred = best_model.predict(X_test)
y_probability = best_model.predict_proba(X_test)[:, 1]

print("Best Model:", best_name)
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred, zero_division=0))
print("Recall:", recall_score(y_test, y_pred, zero_division=0))
print("F1 Score:", f1_score(y_test, y_pred, zero_division=0))
print("ROC-AUC:", roc_auc_score(y_test, y_probability))
print("Average Precision:", average_precision_score(y_test, y_probability))

print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))
```

```python
ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred,
    display_labels=["No", "Yes"]
)

plt.title("Confusion Matrix")
plt.show()
```

```python
sample_size = min(2000, len(X_test))

X_sample = X_test.sample(
    n=sample_size,
    random_state=RANDOM_STATE
)

y_sample = y_test.loc[X_sample.index]

importance = permutation_importance(
    best_model,
    X_sample,
    y_sample,
    scoring="roc_auc",
    n_repeats=3,
    random_state=RANDOM_STATE,
    n_jobs=1
)

importance_df = pd.DataFrame({
    "Feature": X.columns,
    "Importance": importance.importances_mean
}).sort_values(
    by="Importance",
    ascending=False
)

importance_df.head(15)
```

```python
%%writefile predict.py

import joblib
import pandas as pd

from fastapi import FastAPI
from pydantic import BaseModel

app = FastAPI(title="Bank Marketing Prediction API")

model = joblib.load("model.bin")


class CustomerInput(BaseModel):
    age: int
    job: str
    marital: str
    education: str
    default: str
    balance: int
    housing: str
    loan: str
    contact: str
    day: int
    month: str
    campaign: int
    pdays: int
    previous: int
    poutcome: str


@app.get("/")
def home():
    return {"message": "Bank Marketing Prediction API is running"}


@app.get("/health")
def health():
    return {"status": "healthy"}


@app.post("/predict")
def predict(customer: CustomerInput):

    data = pd.DataFrame([customer.model_dump()])

    prediction = int(model.predict(data)[0])
    probability = float(model.predict_proba(data)[0][1])

    return {
        "prediction": prediction,
        "subscription": "yes" if prediction == 1 else "no",
        "probability_yes": round(probability, 4)
    }
```

```python
%%writefile train.py

import os
import zipfile
import urllib.request
import pandas as pd
import joblib

from sklearn.model_selection import GridSearchCV, StratifiedKFold
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

url = "https://archive.ics.uci.edu/static/public/222/bank+marketing.zip"

os.makedirs("data", exist_ok=True)

zip_path = "data/bank-marketing.zip"
urllib.request.urlretrieve(url, zip_path)

def find_csv(zip_data):
    from io import BytesIO

    with zipfile.ZipFile(zip_data) as z:
        for name in z.namelist():
            if os.path.basename(name).lower() == "bank-full.csv":
                return z.read(name)

        for name in z.namelist():
            if name.lower().endswith(".zip"):
                result = find_csv(BytesIO(z.read(name)))
                if result is not None:
                    return result

    return None

from io import BytesIO

with open(zip_path, "rb") as f:
    csv_data = find_csv(BytesIO(f.read()))

if csv_data is None:
    raise FileNotFoundError("bank-full.csv not found")

df = pd.read_csv(BytesIO(csv_data), sep=";")

df = df.drop(columns=["duration"])
df["y"] = df["y"].map({"yes": 1, "no": 0})

X = df.drop(columns=["y"])
y = df["y"]

numeric_features = X.select_dtypes(include=["number"]).columns.tolist()
categorical_features = X.select_dtypes(include=["object"]).columns.tolist()

preprocessor = ColumnTransformer([
    ("numeric", Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]), numeric_features),

    ("categorical", Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(handle_unknown="ignore"))
    ]), categorical_features)
])

models = [
    (
        LogisticRegression(max_iter=2000, class_weight="balanced"),
        {"model__C": [0.1, 1, 10]}
    ),
    (
        DecisionTreeClassifier(random_state=42, class_weight="balanced"),
        {"model__max_depth": [5, 10, 20]}
    ),
    (
        RandomForestClassifier(
            random_state=42,
            class_weight="balanced_subsample",
            n_jobs=-1
        ),
        {"model__n_estimators": [100, 200],
         "model__max_depth": [10, 20]}
    )
]

cv = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)

best_score = 0
best_model = None

for model, params in models:
    pipeline = Pipeline([
        ("preprocessor", preprocessor),
        ("model", model)
    ])

    grid = GridSearchCV(
        pipeline,
        params,
        scoring="roc_auc",
        cv=cv,
        n_jobs=1
    )

    grid.fit(X, y)

    if grid.best_score_ > best_score:
        best_score = grid.best_score_
        best_model = grid.best_estimator_

joblib.dump(best_model, "model.bin")

print("Training completed.")
print("Best CV ROC-AUC:", best_score)
```

```python
%%writefile README.md

# Bank Marketing Subscription Prediction

## Problem Description

This project predicts whether a bank customer will subscribe to a term deposit based on customer information and previous marketing campaign data.

The problem is a binary classification task.

## Dataset

The dataset is Bank Marketing from the UCI Machine Learning Repository.

Dataset link:
https://archive.ics.uci.edu/dataset/222/bank+marketing

The dataset contains 45,211 records and 17 original features.

The duration feature was removed because it is only known after a phone call and would cause data leakage.

## Data Preparation

- Missing values were handled using SimpleImputer.
- Numerical features were standardized.
- Categorical features were encoded using OneHotEncoder.
- Data was divided into training and testing sets.

## Models

The following models were evaluated:

- Logistic Regression
- Decision Tree
- Random Forest

GridSearchCV and StratifiedKFold were used for hyperparameter tuning and model selection.

ROC-AUC was used as the main evaluation metric.

## API

The prediction service was developed using FastAPI.

Available endpoints:

- GET /
- GET /health
- POST /predict

## Run Locally

Install dependencies:

pip install -r requirements.txt

Start the API:

uvicorn predict:app --host 0.0.0.0 --port 8000

Open:

http://127.0.0.1:8000/docs

## Docker

Build the image:

docker build -t bank-marketing-api .

Run the container:

docker run -p 8000:8000 bank-marketing-api

## Files

- notebook.ipynb: Data analysis, training and evaluation
- train.py: Model training script
- predict.py: FastAPI prediction service
- model.bin: Trained machine learning pipeline
- requirements.txt: Project dependencies
- Dockerfile: Docker configuration
```
